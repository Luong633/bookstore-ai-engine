```python
import time
import math
import random
import numpy as np

# ==========================================
# 1. CÁCH LÀM CỦA PYTHON THUẦN (VÒNG LẶP FOR)
# ==========================================
def cosine_similarity_python(query_vec, db_vectors):
    results = []
    # Tính độ dài của vector truy vấn (L2 Norm)
    norm_query = math.sqrt(sum(q * q for q in query_vec))
    
    # Lặp qua từng vector trong cơ sở dữ liệu
    for db_vec in db_vectors:
        # Tích vô hướng (Dot Product)
        dot_product = sum(q * v for q, v in zip(query_vec, db_vec))
        # Độ dài của vector trong DB
        norm_db = math.sqrt(sum(v * v for v in db_vec))
        
        # Tính Cosine
        cosine = dot_product / (norm_query * norm_db)
        results.append(cosine)
    return results

# ==========================================
# 2. (NUMPY VECTORIZATION)
# ==========================================
def cosine_similarity_numpy(query_vec, db_matrix):
    # Tính dot product cùng lúc với 100.000 vector bằng phép nhân ma trận
    dot_products = np.dot(db_matrix, query_vec)
    
    # Tính độ dài của query và của toàn bộ 100.000 vector trong DB
    norm_query = np.linalg.norm(query_vec)
    norms_db = np.linalg.norm(db_matrix, axis=1)
    
    # Phép chia broadcasting (chia mảng cho mảng cực nhanh)
    return dot_products / (norm_query * norms_db)

# ==========================================
# 3. BENCHMARK (ĐO LƯỜNG TỐC ĐỘ THỰC TẾ)
# ==========================================
print("Đang tạo bộ dữ liệu giả lập...")
DIMENSIONS = 384 # Số chiều vector chuẩn của mô hình ngôn ngữ nhỏ
NUM_VECTORS = 100_000 # 100 nghìn cuốn sách trong DB

# Tạo dữ liệu ngẫu nhiên
query_python = [random.random() for _ in range(DIMENSIONS)]
db_python = [[random.random() for _ in range(DIMENSIONS)] for _ in range(NUM_VECTORS)]

# Chuyển đổi sang định dạng mảng C của NumPy
query_numpy = np.array(query_python, dtype=np.float32)
db_numpy = np.array(db_python, dtype=np.float32)

print(f"Bắt đầu so sánh tìm kiếm 1 câu hỏi với {NUM_VECTORS} cuốn sách (Vector {DIMENSIONS} chiều)...\n")

# Đo tốc độ Python thuần
start_time = time.time()
res_python = cosine_similarity_python(query_python, db_python)
python_time = time.time() - start_time
print(f"⏱ Tốc độ Python thuần (vòng lặp FOR) : {python_time:.4f} giây")

# Đo tốc độ NumPy
start_time = time.time()
res_numpy = cosine_similarity_numpy(query_numpy, db_numpy)
numpy_time = time.time() - start_time
print(f"⏱ Tốc độ NumPy (Vectorization)        : {numpy_time:.4f} giây")

speedup = python_time / numpy_time
print("-" * 50)
print(f" KẾT LUẬN: NumPy tính toán nhanh gấp {speedup:.1f} lần so với Python thuần!")
```
